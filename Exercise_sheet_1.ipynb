## Exercise 1 - FizzBuzz

Write a program that prints integer numbers from 1 to 25. For multiples of 3 print `"Fizz"` instead of the number, and for multiples of 5 print `"Buzz"`.

**Careful:** for multiples of 3 and 5, you should print `"FizzBuzz"`!

```python
for i in range(1, 26):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
```

## Exercise 2 - Leap years

**(a)** Write a function that, given a year, reports if it is a [leap year](https://en.wikipedia.org/wiki/Leap_year). The tricky thing is that a leap year in the Gregorian calendar occurs:
 * on every year that is evenly divisible by 4
 * except every year that is evenly divisible by 100
 * unless the year is also evenly divisible by 400
 
For example, 1995 is not a leap year, but 1996 is. 1900 is not a leap year, but 2000 is.

**(b)** Test your function on some selected years that allow you to check if it provides the correct solution.

```python
def leapyear(year):
    """ 
    returns True or False depending on the input year being a leapyear or not
    As there was no consistent rule before 45 BC (Julian calendar reform, by Gaius Iulius Caesar as Pontifex Maximus)
    None is returned for years before -45. The Julian calender had EVERY year divisible by 4 as leapyear.
    Only in 1582 (and strictly speaking, starting in that year, and then every territory following its own schedule)
    the Gregorian reform was implemented. We ignore this territorial aspect. You find the details here: https://de.wikipedia.org/wiki/Gregorianischer_Kalender
    For example, the archduchy Austria implemented the calendar in 1584, The UK in 1752, France 1806, the Soviet Union 1918 and Greece (except Athos) 1923
    Apparently, Athos never implemented it.
    """
    if year < -45:                                                  # The Julian calendar was introduced 45 BC. There was no consistent leap year before
        return None
    if year < 1582 and year % 4 == 0:                               # with the Julian reform, every year divisible by 4 was a leap year
        return True
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):    # with Gregorian reform, we got the more accurate rule
        return True                                                 # to be more precise, one would have to check for the region
    else:                                                           # because the calendar was deployed at different dates in different territories
        return False

testyears = [-100, 0, 100, 1200, 1500, 1600, 1700, 1704, 1900, 1964, 2000, 2004, 2026]
for year in testyears:
    if leapyear(year) == None:
        print (f'The year {year} cannot be ascertained as being a leapyear or not')
    elif leapyear(year):
        print(f'The year {year} is a leapyear')
    else:
        print(f'The year {year} is NOT a leapyear')
```

## Exercise 3 - Rigged dice

**(a)** Create a rigged dice function that 25% of the time returns the number 6. The rest of the time it returns the integers 1,2,3,4,5 uniformly.

**(b)** Test your function, by calling it 100k times and checking that 6 is returned in the range of 24700-25300 (inclusive) times. You do not need to check that numbers 1 to 5 are returned uniformly or randomly, but you need to check that your function returns integers in the range 1-6 (inclusive). Such a test works roughly 98% of the time, so you might want to run it twice if you are confident your solution is correct.

```python
import random

def roll_die():
    """ 
    Roll a die once. Return the result. The die is rigged, it will return a six
    with 25% probability. The numbers 1-5 are equally probable
    """
    a = random.random()
    if a <= 0.25:
        return 6
    else:
       return random.randint(1,5) 

sixes = 0
others = 0
wrongs = 0
for i in range(100_000):
    result = roll_die()
    match result:
        case 6:
            sixes += 1
        case 1 | 2 | 3 | 4 | 5:
            others += 1
        case _:
            wrongs += 1

if sixes >= 24700 & sixes <= 25300:
    expected = ', as expected'
else:
    expected = ', which was NOT expected'
print(f'I rolled {sixes} times the six{expected}')
print(f'I rolled {others} times 1-5')
print(f'I rolled {wrongs} times another number')
```

## Exercise 4 - Find the fake coin

Consider this famous puzzle: _"You have 8 coins: all the coins have equal weight, except one. The odd one weights less than the others, not being of pure gold. In the worst case, how many iterations do you need to find the odd one out on a two plates scale?"_

Write a function that returns the minimum amount of weighing it will take to measure `N` coins in every possible occurrence (including worst case scenario, i.e.: without relying on luck at all!). The number `N` is guaranteed to be a positive integer.

**Tip:** being able to think recursively might help here!

**Note:** this is clearly a logical problem, because a good coder is a master of pattern recognition and subsequent optimization! ;)

```python
""" 
When we are left with one coin, we are finished.
When we have more coins, we divide them into three piles with equal number of coins
We weigh two piles, and that tells us, where the fake coin is
1. equal piles -> coin is in third pile
2. unequal piles -> coin is in lighter pile
However, we need to take care of the remaining coins, when the total number
is not divisible by 3 without remainder
"""

def min_weighing(N):
    if N <= 1:
        return 0
    else:
        return 1 + min_weighing((N + 2) // 3)       # we add one weighing

for coins in range(1, 100):
    print(f'for {coins} coins {min_weighing(coins)} weighings are needed')
```

```python
# Elliot's implementation
""" 
No need to split the check into if/then sections, 
we can simply evaluate the term (health > 0), which will result 
in True or False.
I did not pay special attention to the aspect of health being an integer between -10 and +10
But, as we can rely on it, it is not necessary to confirm the type in the check_alive function.
The assert statement is specifically made for unit tests. The "is" operator ensures that the returned
object is identical to True/False, not just compatible (would check_alive return 1 and 0 instead of True False
it would still be possible to use it as boolean, and it would be the same sematic, but it would NOT be
the same object). The unit test ascertains that the input values -11..0 result in False, and the 
input values 1..11 result in True. In case of a failed assert, an error is thrown. 
"""
def check_alive(health):
    return health > 0

for h in range(-11, 1):
    assert check_alive(h) is False, f'wrong health at {h}, should be False but is  {check_alive(h)}'

for h in range(1, 12):
    assert check_alive(h) is True, f'wrong health at {h}, should be True, but is {check_alive(h)}'

print('all ok')
```
