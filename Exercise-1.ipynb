## Exercise 1 - FizzBuzz

Write a program that prints integer numbers from 1 to 25. For multiples of 3 print `"Fizz"` instead of the number, and for multiples of 5 print `"Buzz"`
    
**Careful:** for multiples of 3 and 5, you should print `"FizzBuzz"`!

```python
numbers = []

for number in range(1, 31):
    if number % 15 == 0:
        numbers.append('FizzBuzz')
    elif number % 5 == 0:
        numbers.append('Buzz')
    elif number % 3 == 0:
        numbers.append('Fizz')
    else:
        numbers.append(number)

numbers
```

```python
def list_version(numbers):
    result = []
    for number in numbers:
        if number % 15 == 0:
            result.append("FizzBuzz")
        elif number % 3 == 0:
            result.append("Fizz")
        elif number % 5 == 0:
            result.append("Buzz")
        else:
            result.append(number)
    return result


def mask_version(numbers):
    fizz = numbers % 3 == 0
    buzz = numbers % 5 == 0

    result = numbers.astype(object)
    result[fizz] = "Fizz"
    result[buzz] = "Buzz"
    result[fizz & buzz] = "FizzBuzz"
    return result


def select_version(numbers):
    return np.select(
        [
            numbers % 15 == 0,
            numbers % 3 == 0,
            numbers % 5 == 0,
        ],
        ["FizzBuzz", "Fizz", "Buzz"],
        default=numbers.astype(str),
    )


numbers = np.arange(1, 1_000_001)

%timeit mask_version(numbers)
%timeit select_version(numbers)
%timeit list_version(numbers)
```

## Exercise 2 - Leap years

 **(a)** Write a function that, given a year, reports if it is a [leap year](https://en.wikipedia.org/wiki/Leap_year). The tricky thing is that a leap year in the Gregorian calendar occurs:
* on every year that is evenly divisible by 4
* except every year that is evenly divisible by 100
* unless the year is also evenly divisible by 400
  
For example, 1995 is not a leap year, but 1996 is. 1900 is not a leap year, but 2000 is.
   
**(b)** Test your function on some selected years that allow you to check if it provides the correct solution.

```python
def is_leap(year):
    if year % 4 == 0:
        if year % 100 == 0:
            if year % 400 == 0:
                return True
            else:
                return False
        else:
            return True
    else:
        return False
    
print(f"2020 is a leap year: {is_leap(2020)}")  # True
print(f"1900 is a leap year: {is_leap(1900)}")  # False   
print(f"2000 is a leap year: {is_leap(2000)}")  # True
```

## Exercise 3 - Rigged dice

**(a)** Create a rigged dice function that 25% of the time returns the number 6. The rest of the time it returns the integers 1,2,3,4,5 uniformly.

**(b)** Test your function, by calling it 100k times and checking that 6 is returned in the range of 24700-25300 (inclusive) times. You do not need to check that numbers 1 to 5 are returned uniformly or randomly, but you need to check that your function returns integers in the range 1-6 (inclusive). Such a test works roughly 98% of the time, so you might want to run it twice if you are confident your solution is correct.
   

```python
import numpy as np

def dice():
    if np.random.random() < 0.25:
        return 6
    else:
        return np.random.randint(2, 6)

sixes = one_to_five = other = 0

for _ in range(100_000):
    result = dice()
    if result == 6:
        sixes += 1
    elif result in [1, 2, 3, 4, 5]:
        one_to_five += 1
    else:
        other += 1

print(f"Sixes: {sixes}, Ones to Fives: {one_to_five}, Others: {other}")
```

## Exercise 4 - Find the fake coin

"Consider this famous puzzle: "You have 8 coins: all the coins have equal weight, except one. The odd one weights less than the others, not being of pure gold. In the worst case, how many iterations do you need to find the odd one out on a two plates scale?
  
Write a function that returns the minimum amount of weighing it will take to measure `N` coins in every possible occurrence (including worst case scenario, i.e.: without relying on luck at all!). The number `N` is guaranteed to be a positive integer.
  
**Tip:** being able to think recursively might help here!

**Note:** this is clearly a logical problem, because a good coder is a master of pattern recognition and subsequent optimization! 

```python
"""
The following function calculates the minimum number of weighings needed to find a counterfeit coin among n coins using a balance scale. 
The function uses a recursive approach, dividing the coins into three groups and weighing them against each other. 
The base case is when there is one or no coins left, in which case no weighings are needed. For larger numbers of coins, 
the function adds one weighing and recursively calls itself with the number of coins reduced to approximately one-third of the original number.
the (n + 2) // 3 expression ensures that the number of coins is rounded up to the nearest integer when dividing by 3, and is equivalent to math.ceil(n / 3). 
This is important because if there are leftover coins after dividing by 3, they still need to be considered in the next weighing.
The result is the same as  given by the formula ceil(log3(n)).
"""

from math import ceil, log

def min_weighings(n):
    if n <= 1:
        return 0

    return 1 + min_weighings((n + 2) // 3)

for n in range(1, 100):
    print(f"N: {n}: min weighings (recursive): {min_weighings(n)}, min weighings (log3): {ceil(log(n, 3))}")
```
